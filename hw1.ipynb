1. 设计一个算法来计算你所能获取的最大利润和对应买入和卖出的日期。请分析算法方案, 计算其时间复杂度, 并且使用python编程实现该算法。

2. \* 设计一个时间复杂度为 $O(n)$的算法实现该算法

e.g. :
---
profits=[3,2,1,-7,5,2,-1,3,-1], 第5天买入, 第8天卖出, 收益最大：9

idea: 算法方案： 遍历数组profits，两次循环枚举计算所有的买入日期和卖出日期的利润，然后找出最大利润，并记录对应日期，最后返回变量。

````python
# add your code here
#
def max_profits(profits):
    max_sum = profits[0]
    buy_day = 0
    sell_day = 0

    for i in range(len(profits)):
        current_sum = 0
        for j in range(i, len(profits)):
            current_sum += profits[j]

            if current_sum > max_sum:
                max_sum = current_sum
                buy_day = i
                sell_day = j
    return max_sum, buy_day + 1, sell_day + 1


profits = [3, 2, 1, -7, 5, 2, -1, 3, -1]

max_profits, buy_day, sell_day = max_profits(profits)

print("买入日期:", buy_day)
print("卖出日期:", sell_day)
print("最大收益:", max_profits)

# your algorithm time complexity is:
#代码中有两个for循环，遍历profits数组找到最大利润和，所以时间复杂度为O(n^2)
````

idea: 满足$O(logn)$的要求，因此使用二分法来设计算法，每次取中间位置，再和左右元素比较 ，如果left<min或right>min,则局部最小值在左边或右边，更新下标，直到找到满足要求的局部最小值（查找资料二分法思路）。

````python
# add your code here
def local_min(A):
    min =0
    left=1
    right=len(A)-2

    while left<=right:
        min=(left +right)//2
        if A[min - 1] > A[min] and A[min] < A[min + 1]:
            return A[min], min
        elif A[min - 1] < A[min]:
            right= min -1
        else:
            left= min +1
    return None


A = [9, 3, 7, 2, 1, 4, 5]
value, index = local_min(A)

print("局部最小值:", value)
print("位置:", index)
# your algorithm time complexity is:
````

3. \* 假设数组中包含的数据总数目超过了计算机的存储能力, 请设计一个算法, 找到这堆数据的前k小的数值, 计算时间复杂度, 并使用python实现该算法, 假设计算机一定能存储k个数据。

e.g.：
---
数组arr=[5,4,3,2,6,1,88,33,22,107] 的前3个最小数据为：1, 2, 3

idea:

1.利用冒泡排序先升序排列数组arr，再直接输出前k个元素即可。算法时间复杂度为$O(n^2)$

2.先建立一个最小堆，把数组中的n个元素依次插入到空堆中，每次插入后通过向上调整保持最小堆性质，使堆顶始终是当前所有元素中的最小值。建好堆之后，连续执行k次 Extract-min操作，每次取出堆顶元素，也就是当前剩余元素中的最小值，并通过向下调整重新维护堆结构。由于最小堆每次弹出的都是当前最小元素，所以连续弹出k次得到的结果就是从小到大排列的最小k个数。总时间复杂度为 O(nlogn)。（借助大模型）

````python
# add your code here
def bubble_sort(arr, k):
    n=len(arr)
    for i in range(n):
        swap =False
        for j in range(0,n-1):
            if arr[j]>arr[j+1]:
                arr[j],arr[j+1]=arr[j+1],arr[j]
                swap= True
        if not swap:
            break
    return arr[:k]

arr = [5, 4, 3, 2, 6, 1, 88, 33, 22, 107]
k = 3

print(bubble_sort(arr, k))
# your algorithm time complexity is:O(n^2)
````

````python
def heappush(heap, x):
    """
    向最小堆中插入一个元素 x
    """
    heap.append(x)
    i = len(heap) - 1

    # 向上调整
    while i > 0:
        parent = (i - 1) // 2

        if heap[parent] <= heap[i]:
            break

        heap[parent], heap[i] = heap[i], heap[parent]
        i = parent


def heappop(heap):
    """
    从最小堆中弹出堆顶最小值
    """
    if not heap:
        raise IndexError("heap is empty")

    min_value = heap[0]

    # 把最后一个元素放到堆顶
    last = heap.pop()

    if heap:
        heap[0] = last

        # 向下调整
        i = 0
        n = len(heap)

        while True:
            left = 2 * i + 1
            right = 2 * i + 2
            smallest = i

            if left < n and heap[left] < heap[smallest]:
                smallest = left

            if right < n and heap[right] < heap[smallest]:
                smallest = right

            if smallest == i:
                break

            heap[i], heap[smallest] = heap[smallest], heap[i]
            i = smallest

    return min_value


def k_smallest_by_min_heap(arr, k):
    heap = []

    # 一个一个插入空堆
    for x in arr:
        heappush(heap, x)

    result = []

    # 执行 k 次 Extract-min
    for _ in range(k):
        result.append(heappop(heap))

    return result


arr = [5, 4, 3, 2, 6, 1, 88, 33, 22, 107]
k = 3

print(k_smallest_by_min_heap(arr, k))
````

## 问题 6

> 星球碰撞问题：现有n个星球，在同一条直线上运行，如数组A所示，元素的绝对值表示星球的质量，负数表示星球自右向左运动，正数表示星球自左向右运动，当两个星球相撞的时候，质量小的会消失，大的保持不变，**质量相同的两个星球碰撞后自右向左运动的星球消失，自左向右的星球保持不变**，假设所有星球的速度大小相同。
>
> $ A=[23,-8, 9, -3, -7, 9, -23, 22] $

请设计一个算法模拟星球的运行情况，输出最终的星球存续情况（输出一个数组），分析算法设计思路，计算时间复杂度，并基于python编程实现。

e.g.
---
1.  输入： A=[-3,-6,2,8, 5,-8,9,-2,1]， 输出：[-3, -6, 2, 8, 9, 1]

2. 输入：A=[23,-8, 9, -3, -7, 9, -23, 22], 输出：[23, 22]

idea: 
两个星球碰撞由于质量会消失，而且自右向左运动消失，自左向右不变，与栈结构相似（栈顶向右），用一个栈保存当前还存活的星球。
创建一个空列表 stack 来模拟栈，遍历数组 A 中的每一个星球 planet，如果 planet > 0（向右飞），它目前不会和任何东西撞，直接把它加入栈；如果 planet < 0（向左飞），需要看栈顶的星球：1如果栈顶质量<当前星球的质量，stack.pop() 出栈。继续用当前星球去撞下一个栈顶。2如果栈顶质量>当前星球的质量：当前星球被撞，结束判定。3如果栈顶质量=当前星球的质量，正数赢，负数消失。当前星球被撞，结束判定。4如果没有发生碰撞（栈为空，或者栈顶也是负数），把它加入栈：stack.append(planet)。
（借助大模型）

````python
# add your code here
def planet_collision(A):
    stack = []

    for planet in A:
        # 只有当遇到负数（向左移动），且栈顶是正数（向右移动）时，才会发生碰撞
        # 用一个标志位记录当前负数星球是否还存活
        alive = True

        while alive and planet < 0 and stack and stack[-1] > 0:
            # 发生碰撞，比较质量（绝对值）
            mass = abs(planet)
            top_mass = stack[-1]

            if top_mass < mass:
                # 栈顶正数被撞碎，弹出栈顶，负数继续下一轮碰撞判断
                stack.pop()
            elif top_mass > mass:
                # 负数被撞碎，标记死亡
                alive = False
            else:
                # 质量相等：规则说“向左的消失，向右的不变” -> 也就是负数碎掉，正数保留
                alive = False

        # 如果经过一系列碰撞后，当前星球还存活（或者它一开始就是正数），将其压入栈
        if alive:
            stack.append(planet)

    return stack


A1 = [-3, -6, 2, 8, 5, -8, 9, -2, 1]
print(f"eg.1结果: {planet_collision(A1)}")

A2 = [23, -8, 9, -3, -7, 9, -23, 22]
print(f"eg.2结果: {planet_collision(A2)}")
# your algorithm time complexity is: O(n)
````

> 0-1背包问题：现在有4块大理石，每个大理石对应的重量和价值使用一个元组表示, 即（重量，价值），4块大理石的重量和价值为：[(5,10), (4,40), (6,30), (3, 50)], 假设你有一辆最大承重为9的小推车，请问使用这个小推车装大理石的最大价值为多少

请设计一个算法解决该问题，分析算法设计思路，计算时间复杂度，并基于python编程实现

idea:
利用动态规划的思想，构建一个一维数组记录小推车在各个承重阶段下的最大价值。对每一块大理石，比较当前容量下的已有最大价值，与腾出该大理石重量后剩余容量的最大价值加上该大理石自身价值之和，取两者中的较大值进行状态更新，同时为了保证每块大理石最多只被选择一次且优化内存空间，算法采取了逆序遍历推车容量的方式，最终数组末尾计算出的值即为全局最优解。时间复杂度为$O(N*W)$,其中N为大理石数量，W为推车最大容量。（借助大模型）

````python
# add your code here
def knapsack_01(items, capacity):
    dp = [0] * (capacity + 1)
    
    for weight, value in items:
        for j in range(capacity, weight - 1, -1):
            dp[j] = max(dp[j], dp[j - weight] + value)
    return dp[capacity]

marbles = [(5, 10), (4, 40), (6, 30), (3, 50)]
max_capacity = 9
max_value = knapsack_01(marbles, max_capacity)
print(f"小推车装大理石的最大价值为: {max_value}") 

# your algorithm time complexity is: O(N*W) N为大理石数量，W为推车最大容量
````

2.

 <div style="text-align:center;width:200px;"> <img alt="" src="./fig/hw3q9-1.jpg"> </div>
输入： root=[3,1,4,null,null,2]， 输出为：[2,1,4,null,null,3]

解释：2 不能在 3 的右子树中，因为 2 < 3 。交换 2 和 3 使二叉搜索树有效。

idea:

利用二叉搜索树中序遍历结果应该是升序序列这一性质。因为题目中只有两个节点的值被错误交换，所以对树进行中序遍历时，一定会出现逆序现象，即前一个节点值prev.val > 当前节点值node.val。第一次发现逆序时，把较大的节点记录为first，把较小的节点记录为second。如果后面又发现一次逆序，则继续更新second为当前较小的节点。遍历结束后，只需要交换first和second的值，就能在不改变树结构的情况下恢复原来的二叉搜索树。该算法访问每个节点一次，时间复杂度为 O(n)。（借助大模型）

````python
# add your code here
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right


def recoverTree(root):
    first = None
    second = None
    prev = None

    def inorder(node):
        nonlocal first, second, prev

        if node is None:
            return

        inorder(node.left)

        if prev is not None and prev.val > node.val:
            if first is None:
                first = prev
            second = node

        prev = node

        inorder(node.right)

    inorder(root)

    if first is not None and second is not None:
        first.val, second.val = second.val, first.val

    return root


def level_order(root):
    """
    层序遍历输出二叉树，方便查看结果
    """
    if root is None:
        return []

    queue = [root]
    result = []

    while queue:
        node = queue.pop(0)

        if node is None:
            result.append(None)
        else:
            result.append(node.val)
            queue.append(node.left)
            queue.append(node.right)

    # 去掉末尾多余的 None
    while result and result[-1] is None:
        result.pop()

    return result


# 构造输入 root = [1,3,None,None,2]
# root = TreeNode(1)
# root.left = TreeNode(3)
# root.left.right = TreeNode(2)

# 构造输入 root = [3,1,4,None,None,2]
root = TreeNode(3)
root.left = TreeNode(1)
root.right = TreeNode(4)
root.right.left = TreeNode(2)

print("恢复前：", level_order(root))

recoverTree(root)

print("恢复后：", level_order(root))

# your algorithm time complexity is:O(n)
````

使用深度优先搜索DFS从根节点遍历到每一个叶子节点，在遍历过程中维护当前路径已经形成的数字。每到一个新节点，就把原来的数字乘以10再加上当前节点值，即current= current*10 + node.val。当遍历到叶子节点时，说明已经形成了一条完整的根到叶子路径数字，就把这个数字返回或累加到总和。最后把所有叶子路径形成的数字加起来。
该算法每个节点只访问一次，时间复杂度是 O(n)。（借助大模型）

````python
# add your code here
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right


def sum_numbers(root):
    def dfs(node, current):
        if node is None:
            return 0

        current = current * 10 + node.val
        if node.left is None and node.right is None:
            return current
        return dfs(node.left, current) + dfs(node.right, current)

    return dfs(root, 0)

#eg1
# root = TreeNode(1)
# root.left = TreeNode(2)
# root.right = TreeNode(3)

root = TreeNode(4)
root.left = TreeNode(9)
root.right = TreeNode(0)
root.left.left = TreeNode(5)
root.left.right = TreeNode(1)


print(sum_numbers(root))


# your algorithm time complexity is: O(n)
````
